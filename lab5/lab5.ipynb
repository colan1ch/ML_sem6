## Импорт библиотек и загрузка данных

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, f1_score, classification_report

from sklearn.ensemble import (
    BaggingClassifier,
    RandomForestClassifier,
    AdaBoostClassifier,
    GradientBoostingClassifier
)

sns.set_theme(style="whitegrid", palette="muted", font_scale=1.1)
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['figure.dpi'] = 100

import warnings
warnings.filterwarnings('ignore')
```

```python
digits = load_digits()

df = pd.DataFrame(digits.data, columns=digits.feature_names)
df['target'] = digits.target
df['target_name'] = df['target'].map(dict(enumerate(digits.target_names)))

df.head(10)
```

## Подготовка данных

```python
X = df[digits.feature_names].values
y = df['target'].values

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## Обучение моделей

#### Бэггинг

```python
bag_model = BaggingClassifier(n_estimators=100, random_state=42)
bag_model.fit(X_train_scaled, y_train)

y_pred_bag = bag_model.predict(X_test_scaled)
acc_bag = accuracy_score(y_test, y_pred_bag)
f1_bag = f1_score(y_test, y_pred_bag, average='weighted')

print(f'Accuracy: {acc_bag:.4f}')
print(f'F1-score (weighted): {f1_bag:.4f}')
print(classification_report(y_test, y_pred_bag))
```

#### Случайный лес

```python
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train_scaled, y_train)

y_pred_rf = rf_model.predict(X_test_scaled)
acc_rf = accuracy_score(y_test, y_pred_rf)
f1_rf = f1_score(y_test, y_pred_rf, average='weighted')

print(f'Accuracy: {acc_rf:.4f}')
print(f'F1-score (weighted): {f1_rf:.4f}')
print(classification_report(y_test, y_pred_rf))
```

#### AdaBoost

```python
ada_model = AdaBoostClassifier(n_estimators=100, random_state=42)
ada_model.fit(X_train_scaled, y_train)

y_pred_ada = ada_model.predict(X_test_scaled)
acc_ada = accuracy_score(y_test, y_pred_ada)
f1_ada = f1_score(y_test, y_pred_ada, average='weighted')

print(f'Accuracy: {acc_ada:.4f}')
print(f'F1-score (weighted): {f1_ada:.4f}')
print(classification_report(y_test, y_pred_ada))
```

#### Градиентный бустинг

```python
gb_model = GradientBoostingClassifier(n_estimators=100, random_state=42)
gb_model.fit(X_train_scaled, y_train)

y_pred_gb = gb_model.predict(X_test_scaled)
acc_gb = accuracy_score(y_test, y_pred_gb)
f1_gb = f1_score(y_test, y_pred_gb, average='weighted')

print(f'Accuracy: {acc_gb:.4f}')
print(f'F1-score (weighted): {f1_gb:.4f}')
print(classification_report(y_test, y_pred_gb))
```

## Сравнение моделей

```python
results = pd.DataFrame({
    'Модель': ['Бэггинг', 'Случайный лес', 'AdaBoost', 'Градиентный бустинг'],
    'Accuracy': [acc_bag, acc_rf, acc_ada, acc_gb],
    'F1-score (weighted)': [f1_bag, f1_rf, f1_ada, f1_gb]
})

results
```

```python
from sklearn.neural_network import MLPClassifier

mlp = MLPClassifier(hidden_layer_sizes=(1000,1000, 1000, 1000))
mlp.fit(X_train_scaled, y_train)

y_pred_mlp = mlp.predict(X_test_scaled)
acc_mlp = accuracy_score(y_test, y_pred_mlp)
f1_mlp = f1_score(y_test, y_pred_mlp, average='weighted')

print(f'Accuracy: {acc_mlp:.4f}')
print(f'F1-score (weighted): {f1_mlp:.4f}')
print(classification_report(y_test, y_pred_mlp))
```
